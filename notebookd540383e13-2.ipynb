```python
num_withnan_cols = [
    'Sunshine', 'Evaporation', 'Cloud9am', 'Cloud3pm'
]

num_corr15_cols = [
    'Sunshine',
    'Cloud9am',
    'Cloud3pm',
    'MaxTemp',
    'Rainfall',
    'WindGustSpeed',
    'Humidity3pm',
    'Humidity9am',
    'Pressure9am',
    'Pressure3pm',
    'RainToday',
    'Temp3pm',
]

cols_1 = num_cols + cat_cols
cols_2 = num_cols + num_withnan_cols + cat_cols
cols_3 = num_corr15_cols + cat_cols
cols_4 = num_corr90_cols + cat_cols
cols_5 = num_corr90_cols + num_withnan_cols

cols = {
    'without_features': cols_1,
    'with_features': cols_2,
    'less 90': cols_3,
    'more 15': cols_4,
    'less 90 and with_features': cols_5,
}
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer, make_column_selector
from sklearn.model_selection import train_test_split

numeric_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()),
])

preprocessor = ColumnTransformer([
    (
        'num',
        numeric_pipeline,
        make_column_selector(dtype_include='number'),
    ),
    (
        'cat',
        OneHotEncoder(handle_unknown='ignore'),
        make_column_selector(dtype_include=['object', 'category']),
    ),
])
nums_cols = num_cols + num_withnan_cols

X = df[nums_cols + cat_cols]
y = df[target_col].astype(int)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 123, stratify = y)

X_train_final_2 = preprocessor.fit_transform(X_train[cols_2])
X_test_final_2 = preprocessor.transform(X_test[cols_2])

X_train_final_3 = preprocessor.fit_transform(X_train[cols_3])
X_test_final_3 = preprocessor.transform(X_test[cols_3])

```

```python
from sklearn.model_selection import cross_validate
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

pipe = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LogisticRegression(max_iter = 1000))
])
score = {}
for name, col in cols.items():
    scores = cross_validate(pipe, X_train[col], y_train, cv = 5, scoring = ['average_precision', 'f1', 'roc_auc'], return_train_score = True)
    score[name] = {
        metric: f'{values.mean():.4f} ± {values.std():.4f}'
        for metric, values in scores.items()
    }
results = pd.DataFrame(score).T
results = results.drop(columns=['fit_time', 'score_time'])

results
```

Как мы видим по метрикам лучше всего себя показали 2 случая разделения данных: с корреляцией <0.9 и включая признаки, у который пропуски мы заменили средним, для дальнейших вычислений/оценок будем использовать with_features и less 90	

```python
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             confusion_matrix, classification_report, log_loss,
                             roc_auc_score, roc_curve, precision_recall_curve, average_precision_score,
                             ConfusionMatrixDisplay, precision_recall_curve, auc)

model = LogisticRegression(max_iter=1000)
model.fit(X_train_final_2, y_train)

y_proba_log_reg_2 = model.predict_proba(X_test_final_2)[:, 1]

thresholds = np.linspace(0.05, 0.95, 1000)

recall = []
precision = []
F1 = []
tau = []
for t in thresholds:
    y_pred = (y_proba_log_reg_2 >= t).astype(int)

    recall.append(recall_score(y_test, y_pred))
    precision.append(precision_score(y_test, y_pred, zero_division=0))
    F1.append(f1_score(y_test, y_pred))
    
tau.append(thresholds[np.argmax(F1)])

fig, ax = plt.subplots(figsize=(14, 5))

ax.plot(thresholds, recall, label='Recall', color='blue')
ax.plot(thresholds, precision, label='Precision', color='red')
ax.plot(thresholds, F1, label='F1-score', color='purple')

ax.axvline(thresholds[np.argmax(F1)],
           linestyle='--', color='black',
           label='Optimal threshold')

ax.set_xlabel('Threshold')
ax.set_ylabel('Score')
ax.legend()
ax.grid(True)
plt.tight_layout()
plt.show()
```

```python
def pr_auc_score(y_true, y_proba):
    precision, recall, _ = precision_recall_curve(y_true, y_proba)
    return auc(recall, precision)
```

```python
pr_auc_tree = []
f1_tree = []
precision_tree = []
recall_tree = []
depths = np.arange(2, 20)
for i in depths:
    tree = DecisionTreeClassifier(random_state=123, max_depth = i)

    tree.fit(X_train_final_2, y_train)
    
    y_pred_tree = tree.predict(X_test_final_2)
    y_proba_tree = tree.predict_proba(X_test_final_2)[:, 1]
    pr_auc_tree.append(pr_auc_score(y_test, y_proba_tree))
    f1_tree.append(f1_score(y_test, y_pred_tree))
    precision_tree.append(precision_score(y_test,y_pred_tree ))
    recall_tree.append(recall_score(y_test,y_pred_tree ))
fig, ax = plt.subplots(figsize = (14,5))
ax.plot(depths, pr_auc_tree, color = 'green', label = 'PR_AUC')
ax.plot(depths, precision_tree, color = 'red', label = 'Precision')
ax.plot(depths, recall_tree, color = 'blue', label = 'Recall')
ax.plot(depths, f1_tree, color = 'purple', label = 'F1')

ax.set_xlabel('depths')
ax.set_ylabel('Metrics')
ax.legend()
ax.grid(True)

plt.tight_layout()
plt.show()
```

```python
C_list = [0.0001, 0.001, 0.01, 0.1, 1, 10, 100, 1000]
pr_auc_l1 = []
f1_l1 = []
precision_l1 = []
recall_l1 = []
accuracy_l1 = []
roc_auc_l1 = []
for C in C_list:
    model_l1 = LogisticRegression(
    penalty='l1',
    solver='liblinear',
    C=C,
    max_iter=1000)
    model_l1.fit(X_train_final_2, y_train)
    y_proba_l1 = model_l1.predict_proba(X_test_final_2)[:,1]
    y_pred_l1 = model_l1.predict(X_test_final_2)
    
    pr_auc_l1.append(pr_auc_score(y_test, y_proba_l1))
    f1_l1.append(f1_score(y_test, y_pred_l1))
    precision_l1.append(precision_score(y_test, y_pred_l1))
    recall_l1.append(recall_score(y_test, y_pred_l1))
    accuracy_l1.append(accuracy_score(y_test, y_pred_l1))
    roc_auc_l1.append(roc_auc_score(y_test, y_proba_l1))

fig, ax = plt.subplots(figsize = (14,5))
ax.plot(C_list, pr_auc_l1, color = 'green', label = 'PR_AUC')
ax.plot(C_list, precision_l1, color = 'red', label = 'Precision')
ax.plot(C_list, recall_l1, color = 'blue', label = 'Recall')
ax.plot(C_list, f1_l1, color = 'purple', label = 'F1')
ax.plot(C_list, accuracy_l1, color = 'black', label = 'Accuracy')
ax.plot(C_list, roc_auc_l1, color = 'black', label = 'ROC-AUC')

ax.set_xscale('log')
ax.set_xlabel('C')
ax.set_ylabel('Metrics')
ax.legend()
ax.grid(True)

plt.tight_layout()
plt.show()
```
